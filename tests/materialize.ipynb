```python
from datetime import datetime, timedelta
from feast import FeatureStore

# Path to your feature repo
store = FeatureStore(repo_path="../feature_repo")

# Materialize last 30 days (example)
# Start from 2025 - 10 - 1 for consistency with the data generation script
end_date = datetime(2025, 10, 1)
start_date = end_date - timedelta(days=30)

store.materialize(start_date=start_date, end_date=end_date)

print(f"Materialization completed from {start_date} to {end_date}")
```

```python
import pandas as pd

# Retreive online features
entity_rows = pd.DataFrame({
    "pulocationid": [93, 92, 97],
    "event_timestamp": pd.to_datetime([
        "2025-05-30",
        "2025-06-02",
        "2025-06-05",
    ])
}).to_dict(orient="records")

df = store.get_online_features(
    entity_rows=entity_rows,
    features=[
        "green_taxi_features:trip_miles",
        "green_taxi_features:fare_amount",
    ],
).to_df()
```
