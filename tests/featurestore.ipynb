```python
import pandas as pd
from feast import FeatureStore

store = FeatureStore(repo_path="../feature_repo")

entity_df = pd.DataFrame({
    "pulocationid": [93, 92, 97],
    "event_timestamp": pd.to_datetime([
        "2025-05-30",
        "2025-06-02",
        "2025-06-05",
    ])
})

df = store.get_historical_features(
    entity_df=entity_df,
    features=[
        "green_taxi_features:trip_miles",
        "green_taxi_features:fare_amount",
    ],
).to_df()

print(df.head())
```

```python
import pandas as pd
from feast import FeatureStore

store = FeatureStore(repo_path="../feature_repo")

entity_df = pd.DataFrame({
    "pulocationid": [168, 221],
    "event_timestamp": pd.to_datetime([
        "2025-10-01",
        "2025-10-15",
    ])
})

df = store.get_historical_features(
    entity_df=entity_df,
    features=[
        "green_taxi_features:fare_amount",
        "yellow_taxi_features:fare_amount",
        "fhvhv_features:fare_amount",
    ],
    full_feature_names=True,
).to_df()

print(df.head())
```
